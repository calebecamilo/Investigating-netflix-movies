## Os dados
### **netflix_data.csv**
| Column | Description |
|--------|-------------|
| `show_id` | ID do programa |
| `type` | Tipo de programa |
| `title` | Título do programa |
| `director` | Diretor(a) do programa |
| `cast` | Elenco do programa |
| `country` | País de origem |
| `date_added` | Data de inclusão na Netflix |
| `release_year` | Ano de lançamento na Netflix |
| `duration` | Duração do programa em minutos |
| `description` | Descrição do programa |
| `genre` | Gênero do programa |

```python
# Importing pandas and matplotlib
import pandas as pd
import matplotlib.pyplot as plt

# Read in the Netflix CSV as a DataFrame
netflix_df = pd.read_csv("netflix_data.csv")
```

```python
# Start coding here! Use as many cells as you like
subset = netflix_df["release_year"].between(1990,1999)
netflix_df = netflix_df[subset]
duration = netflix_df["duration"].mode().sum()
duration.dtype
plt.hist(netflix_df["duration"])
plt.show()

# Question 2
netflix_df["genre"].value_counts()

# Create a bar plot to vizualize the genres of movies released in the 1990s
netflix_df["genre"].value_counts().plot(kind = "bar", figsize = (16,8), rot = 0)
plt.xlabel("Movie genre")
plt.ylabel("Count of movies")
plt.title("Genre of movies released in the 1990s")

short_action_movies = (netflix_df["genre"] == "Action") & (netflix_df["duration"] < 90)

# Obtainning the total count of movies with short duration
short_movie_count = short_action_movies.sum()
print(short_movie_count)
```
